```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
from sklearn.metrics import mean_squared_error, r2_score
from scipy.stats import pearsonr
```

```python
#данные из методички
h0=np.array([250,230,210,150,90,30])
t0=np.array([12,14.71,14.38,22.67,25.33,65])
v0=np.array((a*3.14/4)*(d**2)*((2*10*h0)**0.5))
v0met=np.array([1.4,1.34,1.28,1.08,0.84,0.48])
```

```python
v0
```

```python
v0=v0.reshape(-1, 1)
v0met=v0met.reshape(-1, 1)
t0=t0.reshape(-1, 1)
```

```python
model0 = linear_model.LinearRegression()
model0.fit(t0,v0)
pred0 = model0.predict(t0)
model0met = linear_model.LinearRegression()
model0met.fit(t0,v0met)
pred0met = model0met.predict(t0)
```

```python
plt.scatter(t0, v0)
plt.plot(t0, pred0)
plt.grid(True)
plt.xlabel("t,c")
plt.ylabel('Vt')
print(f"Коэффициент:{model0.coef_[0]}")
print(f"Свободный член): {model0.intercept_}")
print(f"MSE: {mean_squared_error(t0, pred0)}")
```

```python
plt.scatter(t0, v0met)
plt.plot(t0, pred0met)
plt.grid(True)
plt.xlabel("t,c")
plt.ylabel('Vt')
print(f"Коэффициент:{model0met.coef_[0]}")
print(f"Свободный член): {model0met.intercept_}")
print(f"MSE: {mean_squared_error(t0, pred0met)}")
```

```python
#экспериментальные данные
h=np.array([175,135,95,75,60,30])
t1=np.array([45.53, 49.99, 64, 149, 167, 301])
t2=np.array([47.74,53.21,59.95,146,170,293])
t3=np.array([46.53,52.5,60,140,175,290])
tm=np.array((t1+t2+t3)/3)
sqh=np.array(h**0.5)
a=600
d=0.01
v=np.array((a*3.14/4)*(d**2)*((2*10*h)**0.5))
```

```python
v
```

```python
v=v.reshape(-1, 1)
t1=t1.reshape(-1, 1)
t2=t2.reshape(-1,1)
t3=t3.reshape(-1,1)
tm=tm.reshape(-1,1)
```

```python
model1 = linear_model.LinearRegression()
model2 = linear_model.LinearRegression()
model3 = linear_model.LinearRegression()
modelm = linear_model.LinearRegression()
model1.fit(t1,v)
model2.fit(t2,v)
model3.fit(t3,v)
modelm.fit(tm,v)
pred1 = model1.predict(t1)
pred2 = model2.predict(t2)
pred3 = model3.predict(t3)
predm = modelm.predict(tm)
```

```python
pred3
```

```python
plt.scatter(t1,v)
```

```python
plt.scatter(t1, v)
plt.plot(t1, pred1)
plt.grid(True)
plt.xlabel("t,c")
plt.ylabel('Vt,мл/c')
print(f"Коэффициент:{model1.coef_[0]}")
print(f"Свободный член): {model1.intercept_}")
print(f"MSE: {mean_squared_error(t1, pred1)}")
```

```python
plt.scatter(t2, v)
plt.plot(t2, pred2)
plt.grid(True)
plt.xlabel("t,c")
plt.ylabel('Vt,мл/c')
print(f"Коэффициент:{model2.coef_[0]}")
print(f"Свободный член): {model2.intercept_}")
print(f"MSE: {mean_squared_error(t2, pred2)}")
```

```python
plt.scatter(t3, v)
plt.plot(t3, pred3)
plt.grid(True)
plt.xlabel("t,c")
plt.ylabel('Vt,мл/c')
print(f"Коэффициент:{model3.coef_[0]}")
print(f"Свободный член): {model3.intercept_}")
print(f"MSE: {mean_squared_error(t3, pred3)}")
```

```python
plt.scatter(tm, v)
plt.plot(tm, predm)
plt.grid(True)
plt.xlabel("t,c")
plt.ylabel('Vt,мл/c')
print(f"Коэффициент:{modelm.coef_[0]}")
print(f"Свободный член): {modelm.intercept_}")
print(f"MSE: {mean_squared_error(tm, predm)}")
```

```python
print(f"R²: {r2_score(t1, pred1)}")
print(f"R²: {r2_score(t2, pred2)}")
print(f"R²: {r2_score(t3, pred3)}")
print(f"R²: {r2_score(t0, pred0met)}")
```
